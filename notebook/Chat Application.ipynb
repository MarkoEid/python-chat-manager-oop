## Create a class User with attributes username and messages (list of messages).

## Add methods to:
 * ## Send a message.
* ## View all messages.


```python
class User:
    def __init__(self,username,messages):
        self.username=username
        self.messages=[]

    def send_message(self,msg):
        if msg=="":
            print("you didn't enter the message")
        else:
            self.messages.append(msg)
            print(f"the message is added")
    def view_messages(self):
        print(self.messages)

u1=User("Ahmed",[])
u1.send_message("hello")
u1.view_messages()
```
